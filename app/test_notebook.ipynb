```python
from DocumentLayout import DocumentLayout
from document_context import DocumentContext
from line_classifier import BertLineClassifier
from pathlib import Path

sample_pdf_path = Path("../data/raw/Betalningsintyg.pdf")
layout = DocumentLayout.from_pdf(sample_pdf_path)
context = DocumentContext.from_layout(layout)
classifier = BertLineClassifier()

context.classify_lines(classifier)
```

```python
for item in context.items:
    print(item.text, item.classification)
```
